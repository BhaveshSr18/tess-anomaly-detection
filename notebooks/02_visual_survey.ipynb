```python
import numpy as np
import matplotlib.pyplot as plt

# Load the preprocessed light curve data
X = np.load(r"D:\Astrophysics Project\Astro-Project\data\processed\lightcurves.npy")

print(f"Data shape: {X.shape}")
print(f"Total light curves: {X.shape[0]}, Points per curve: {X.shape[1]}")

# Create figure with subplots
fig, axes = plt.subplots(4, 3, figsize=(14, 10))
axes = axes.flatten()

# Track which indices we've plotted for the legend
plotted_indices = []

for i in range(12):
    # Randomly select a light curve
    idx = np.random.randint(len(X))
    plotted_indices.append(idx)
    
    ax = axes[i]
    ax.plot(X[idx], lw=1.0, color='steelblue', alpha=0.8)
    ax.set_title(f"Light Curve #{idx}", fontsize=10, fontweight='bold')
    ax.set_xlabel("Time Index", fontsize=9)
    ax.set_ylabel("Normalized Flux", fontsize=9)
    ax.grid(True, alpha=0.3, linestyle='--')
    
    # Add some statistics to the plot
    mean_flux = np.mean(X[idx])
    std_flux = np.std(X[idx])
    ax.text(0.98, 0.97, f"μ={mean_flux:.2f}\nσ={std_flux:.2f}", 
            transform=ax.transAxes, fontsize=8, verticalalignment='top',
            horizontalalignment='right', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

fig.suptitle("Random TESS Light Curves - Visual Survey", fontsize=14, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()

print(f"\nDisplayed indices: {plotted_indices}")
```
